```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.columns = df.columns.str.lower()   #here we lowercase the all columns name to read easily.
df.columns = df.columns.str.replace(' ', '_')  # here we raplace that space with " _ " so it makes it easier.
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'}) #here we rename the 'purchase_amount_(usd)' to 'purchase_amount'
```

# Feature Engineering

```python
labels = ['Young_Adult','Adult','Middle_Age','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create new column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
import pandas as pd
from sqlalchemy import create_engine, text
from urllib.parse import quote_plus

# =========================
# 1. MySQL connection details
# =========================

username = "root"
password = "Root@123"   # <-- replace with your actual password
host = "localhost"
port = 3306

# Safely handles @, #, etc. in password
password_encoded = quote_plus(password)

# =========================
# 2. Connect to MySQL server
# =========================

server_engine = create_engine(
    f"mysql+pymysql://{username}:{password_encoded}@{host}:{port}"
)

# =========================
# 3. Create database
# =========================

database = "customer_behavior"

with server_engine.connect() as connection:
    connection.execute(
        text(f"CREATE DATABASE IF NOT EXISTS {database}")
    )
    connection.commit()

print("Database ready! 🎉")

# =========================
# 4. Connect to the database
# =========================

engine = create_engine(
    f"mysql+pymysql://{username}:{password_encoded}@{host}:{port}/{database}"
)

# =========================
# 5. Upload DataFrame to MySQL
# =========================

df.to_sql(
    "customer",
    engine,
    if_exists="replace",
    index=False
)

print("Data uploaded successfully! 🎉")

# =========================
# 6. Read data from MySQL
# =========================

result = pd.read_sql(
    "SELECT * FROM customer LIMIT 5;",
    engine
)

result
```
